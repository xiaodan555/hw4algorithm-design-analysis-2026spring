观察下方的分治算法（divide-and-conquer algorithm）的伪代码, 回答下面问题

```latex
DoSomething(A,p,r)
-----
n := r-p+1
if n=2 and A[p]>A[r] then
    swap A[p] and A[r]
else if n >= 3 then

    DoSomething(A,p,p+m-1)
    DoSomething(A,r-m+1,r)

---
first call: DoSomething(A,1,n)
```

note：$ceil(2n/3)=\left\lceil {2n/3} \right\rceil$；$:=$ 表示赋值, 等价于 $\to$；A是一个包含n的整数元素的数组, 

1. 写出该算法时间复杂度的递归公式, 并求解其对应的渐进表示
2. 描述一下该算法的功能, 并判断是否是最高效的解决方案
3. \* 使用python编程实现上述算法或其对应的更高效版本的算法

idea:

### 1.时间复杂度计算及渐进表示
时间复杂度公式：$T(n)=3T\left(\left\lceil \frac{2n}{3} \right\rceil\right)+\Theta(1)$

渐进表示求解：

根据主定理$T(n)=3T\left(\frac{2n}{3}\right)+\Theta(1)$

得到：$a=3,\qquad b=\frac{3}{2},\qquad f(n)=\Theta(1)$

那么$n^{\log_b a}=n^{\log_{3/2}3}$.

因为$\log_{3/2}3 \approx 2.71$，且$f(n)=\Theta(1)$比$n^{\log_{3/2}3}$小很多，

因此，该算法的时间复杂度为：$T(n)=\Theta\left(n^{\log_{3/2}3}\right) \approx \Theta(n^{2.71})$.

---

### 2.算法功能描述

该算法的功能是对数组进行升序排序。它通过递归地排序前m个数、后m个数、再排序前m个数的方式（m约等于数组长度的2/3），使元素逐步移动到正确位置。

该算法不是最高效的解决方案，因为大多数常见排序算法的时间复杂度都比该算法要小

---

### 3.算法实现

````python
# python实现这个算法
import math

def stooge_like_sort(A):
    def do_something(A, p, r):
        n = r - p + 1

        if n == 2 and A[p] > A[r]:
            A[p], A[r] = A[r], A[p]

        elif n >= 3:
            m = math.ceil(2 * n / 3)

            do_something(A, p, p + m - 1)
            do_something(A, r - m + 1, r)
            do_something(A, p, p + m - 1)

    do_something(A, 0, len(A) - 1)
    return A


A = [4, 1, 6, 2, 5, 3]
print(stooge_like_sort(A))
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:

### 算法设计思路
若第`i`天买入、第`j`天卖出，则总收益为`profits[i] + ... + profits[j]`这样一个连续子数组的和，所以计算最大利润可以转化为找最大连续子数组和，同时记录`i`和`j`

---
### 算法实现

````python
# python编程实现-暴力枚举
def max_profit_bruteforce(profits):
    n = len(profits)
    if n == 0:
        raise ValueError("profits 不能为空")

    max_profit = float("-inf")
    best_buy = best_sell = 0

    for buy in range(n):
        current_sum = 0
        for sell in range(buy, n):
            current_sum += profits[sell]

            if current_sum > max_profit:
                max_profit = current_sum
                best_buy = buy
                best_sell = sell

    return max_profit, best_buy + 1, best_sell + 1

profits=[3,2,1,-7,5,2,-1,3,-1]
max_profit_bruteforce(profits)
````

内层循环枚举卖出日期，最坏情况下也需要遍历`n`次。

因此时间复杂度为：$O(n^2)$

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:

### 算法设计思路

题目给定 $A[1]>A[2]$ 并且 $A[n-1]<A[n]$，这说明数组开头是下降趋势，数组结尾是上升趋势。由于数组中所有元素互不相同，所以相邻元素之间不会相等。

对于数组中的某个元素 $A[i]$，如果它满足 $A[i] < A[i-1]$ 并且 $A[i] < A[i+1]$，则说明 $A[i]$ 比左右两个邻居都小，因此 $A[i]$ 是一个局部最小值。

这里采用**顺序遍历**的方法：

由于局部最小值必须同时有左邻居和右邻居，所以只需要遍历数组中的非边界元素，即从 $A[2]$ 到 $A[n-1]$。

对于每个位置 $i$，判断是否满足 $A[i] < A[i-1]$ 并且 $A[i] < A[i+1]$ ，如果满足，就将 $A[i]$ 加入结果列表。遍历结束后，结果列表中保存的就是数组中所有的局部最小值。

---

### 算法实现

````python
def find_all_local_min(A):
    n = len(A) - 1
    result = []

    for i in range(2, n):
        if A[i] < A[i - 1] and A[i] < A[i + 1]:
            result.append(A[i])

    return result

A = [None, 9, 3, 7, 2, 1, 4, 5]

result = find_all_local_min(A)

print("局部最小元素为:", result)
````

3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:

### 1.排序输出前k个最小：
#### 算法思想

给定一个包含 $n$ 个不同数字的数组 $arr$，要求找到其中最小的 $k$ 个数字，并按从小到大的顺序输出。

这里选择 **归并排序** 实现。

基本步骤如下：

1. 使用归并排序将数组 $arr$ 按从小到大的顺序排序；
2. 排序完成后，数组前 $k$ 个元素就是最小的 $k$ 个数字；
3. 按顺序输出这 $k$ 个数字。

---

#### 时间复杂度计算

归并排序的时间复杂度为：

$
O(n log n)
$

输出前 $k$ 个元素需要：

$
O(k)
$

所以总时间复杂度为：

$
O(n log n + k)
$

由于 $k <= n$，可以写成：

$
O(n log n)
$

---

#### 算法实现

````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


def find_k_smallest(arr, k):
    sorted_arr = merge_sort(arr)
    return sorted_arr[:k]


arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

result = find_k_smallest(arr, k)

print("数组arr={} 的前{}个最小数据为：{}".format(
    arr,
    k,
    "，".join(map(str, result))
))
````

### 2.使用最小堆找前 k 个最小的数字


#### 算法设计思路

给定一个包含 $n$ 个不同数字的数组 $arr$，要求找出其中最小的 $k$ 个数字，并按从小到大的顺序输出。

本题使用 **最小堆** 完成。

最小堆的特点是：`text堆顶元素一定是当前堆中的最小值`

假设某个节点在数组中的下标为 $i$，并且数组下标从 $0$ 开始，那么最小堆的数组下标表示为：

父节点下标：$parent = (i - 1) // 2$

左孩子下标：$left = 2i + 1$

右孩子下标：$right = 2i + 2$

最小堆需要满足每个父节点的值都小于等于它的子节点的值，也就是说$heap[parent] <= heap[child]$

---

#### 建堆过程

题目要求使用“向空堆中不断插入元素”的方式建堆。

假设数组为：$arr = [5, 4, 3, 2, 6, 1]$

从空堆开始，依次插入元素：

```text
1. 插入 5，堆为 [5]

2. 插入 4，先放到最后，得到 [5, 4]
   因为 4 比父节点 5 小，所以 4 上浮，得到 [4, 5]

3. 插入 3，先放到最后，得到 [4, 5, 3]
   因为 3 比父节点 4 小，所以 3 上浮，得到 [3, 5, 4]

4. 插入 2，先放到最后，得到 [3, 5, 4, 2]
   2 比父节点 5 小，交换，得到 [3, 2, 4, 5]
   2 又比父节点 3 小，继续交换，得到 [2, 3, 4, 5]

5. 插入 6，先放到最后，得到 [2, 3, 4, 5, 6]
   6 比父节点 3 大，不需要调整

6. 插入 1，先放到最后，得到 [2, 3, 4, 5, 6, 1]
   1 比父节点 4 小，交换，得到 [2, 3, 1, 5, 6, 4]
   1 又比父节点 2 小，继续交换，得到 [1, 3, 2, 5, 6, 4]
```

所以最终建立好的最小堆为：

```text
[1, 3, 2, 5, 6, 4]
```

对应的堆结构为：

```text
        1
      /   \
     3     2
    / \   /
   5   6 4
```

---

#### Extract-min 操作

`Extract-min` 表示取出堆中的最小值。

因为最小堆的堆顶元素就是最小值，所以每次直接取出堆顶元素即可。取出堆顶后，需要将堆的最后一个元素放到堆顶，然后执行“下沉”操作，重新调整为最小堆。

例如当前堆为：`[1, 3, 2, 5, 6, 4]`

执行第一次 `Extract-min`：

```text
取出堆顶 1

将最后一个元素 4 放到堆顶，得到 [4, 3, 2, 5, 6]

然后比较 4 的左右孩子 3 和 2，选择较小的 2 交换

调整后得到 [2, 3, 4, 5, 6]
```

所以第一次取出的最小值是：

```text
1
```

继续执行 `Extract-min`，就会依次得到：

```text
1，2，3，...
```

因此执行 $k$ 次 `Extract-min`，就可以得到最小的 $k$ 个数字。

**ps：建堆过程和Extract-min操作借助了大模型思考**

---

#### 时间复杂度分析
建堆时，题目要求从空堆开始不断插入元素。

每插入一个元素，最多需要向上调整 $O(\log n)$ 次。

一共插入 $n$ 个元素，所以建堆时间复杂度为：$O(n \log n)$

每次 `Extract-min` 需要进行一次下沉调整，时间复杂度为：$O(\log n)$

执行 $k$ 次 `Extract-min`，时间复杂度为：$O(k \log n)$

所以总时间复杂度为：$O(n \log n + k \log n)$

也可以写成：$O((n + k)\log n)$

---

#### 算法实现

````python
class MinHeap:
    def __init__(self):
        self.heap = []

    # 插入元素
    def insert(self, value):
        self.heap.append(value)
        self.sift_up(len(self.heap) - 1)

    # 上浮调整
    def sift_up(self, index):
        while index > 0:
            parent = (index - 1) // 2

            if self.heap[parent] <= self.heap[index]:
                break

            self.heap[parent], self.heap[index] = self.heap[index], self.heap[parent]
            index = parent

    # 取出堆中的最小值
    def extract_min(self):
        min_value = self.heap[0]

        last_value = self.heap.pop()

        if len(self.heap) > 0:
            self.heap[0] = last_value
            self.sift_down(0)

        return min_value

    # 下沉调整
    def sift_down(self, index):
        n = len(self.heap)

        while True:
            left = 2 * index + 1
            right = 2 * index + 2
            smallest = index

            if left < n and self.heap[left] < self.heap[smallest]:
                smallest = left

            if right < n and self.heap[right] < self.heap[smallest]:
                smallest = right

            if smallest == index:
                break

            self.heap[index], self.heap[smallest] = self.heap[smallest], self.heap[index]
            index = smallest


def find_k_smallest_by_heap(arr, k):
    heap = MinHeap()

    # 从空堆开始，依次插入所有元素
    for num in arr:
        heap.insert(num)

    result = []

    # 执行 k 次 Extract-min
    for _ in range(k):
        result.append(heap.extract_min())

    return result


arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

result = find_k_smallest_by_heap(arr, k)

print("数组arr={} 的前{}个最小数据为：{}".format(
    arr,
    k,
    "，".join(map(str, result))
))
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:

### 算法设计思路
通过观察题目和示例，发现只有当左边的星球向右运动，右边的星球向左运动才会发生碰撞，在数组上就是出现类似$[5,-8]$这种情况会判断哪个元素需要删除，但这种情况不只限于相邻的两个元素，在e.g.2中-23要和前面三个正数都进行比较。

所以这里想到可以用**栈**来存放幸存的星球元素。

大概算法过程：

1.从数组中取出的第一个元素直接放入栈中，

2.接下来每取一个元素都和栈顶进行比较：

- 会出现碰撞的情况：栈顶为正数，该元素为负数（若栈顶绝对值更大，该元素直接删除；若栈顶绝对值更小，栈顶元素取出，该元素和新栈顶比较。
- 其余情况直接将新元素存入栈中。

3.重复2的操作，直到数组中最后一个元素比较完成。

---

### 时间复杂度计算
该算法的时间复杂度为 $O(n)$。

设数组中一共有 $n$ 个星球，算法会从左到右遍历数组，因此外层循环执行 $n$ 次。虽然在处理某些负数星球时会出现 `while` 循环，不断和栈顶元素进行比较，但每个星球最多只会入栈一次，也最多只会被弹出栈一次。

也就是说，整个算法中入栈操作最多执行 $n$ 次，出栈操作最多执行 $n$ 次。因此总操作次数可以表示为 $n+n+n$，所以时间复杂度为 $O(n+n+n)=O(n)$。

---

### 算法实现

````python
def asteroid_collision(A):
    stack = []

    for x in A:
        alive = True

        # 只有栈顶为正数，当前元素为负数时，才会发生碰撞
        while alive and stack and stack[-1] > 0 and x < 0:
            # 栈顶星球质量小，栈顶消失
            if stack[-1] < abs(x):
                stack.pop()

            # 栈顶星球质量大，当前星球消失
            elif stack[-1] > abs(x):
                alive = False

            # 质量相同，题目规定负数方向的星球消失，正数方向的星球保留,所以当前星球消失
            else:
                alive = False

        # 如果当前星球没有消失，就入栈
        if alive:
            stack.append(x)

    return stack


A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
A2 = [23, -8, 9, -3, -7, 9, -23, 22]

print("输入：A={}，输出：{}".format(A1, asteroid_collision(A1)))
print("输入：A={}，输出：{}".format(A2, asteroid_collision(A2)))
````

idea:

### 算法设计思路
本题是典型的0-1背包问题，所以直接考虑用**动态规划**。

#### 状态定义

定义二维数组：$dp[i][j]$，表示在前 i 个大理石中进行选择，当前小推车最大承重为 j 时，能够获得的最大价值

#### 状态转移方程

对于第 $i$ 个大理石，有两种情况：

**1. 不选择第 i 个大理石**

$dp[i][j] = dp[i - 1][j]$

**2. 选择第 i 个大理石**

前提是当前容量 $j$ 能装下它，即：
$j >= weight[i - 1]$

那么：$dp[i][j] = dp[i - 1][j - weight[i - 1]] + value[i - 1]$

所以完整状态转移方程为：
$dp[i][j] = max(dp[i - 1][j], dp[i - 1][j - weight[i - 1]] + value[i - 1])$

如果当前容量装不下第 $i$ 个大理石，则只能不选$dp[i][j] = dp[i - 1][j]$

---

### 时间复杂度计算

设大理石数量为 $n$，小推车最大承重为 $W$。

动态规划需要遍历每个大理石，并且对每个承重容量进行计算。

因此时间复杂度为：$O(nW)$

---

### 算法实现

````python
def knapsack_01(items, W):
    n = len(items)

    dp = [[0 for _ in range(W + 1)] for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight = items[i - 1][0]
        value = items[i - 1][1]

        for j in range(1, W + 1):
            if weight > j:
                dp[i][j] = dp[i - 1][j]
            else:
                dp[i][j] = max(
                    dp[i - 1][j],
                    dp[i - 1][j - weight] + value
                )

    return dp[n][W]


items = [(5, 10), (4, 40), (6, 30), (3, 50)]
W = 9

result = knapsack_01(items, W)

print("大理石信息为：{}".format(items))
print("小推车最大承重为：{}".format(W))
print("能装入大理石的最大价值为：{}".format(result))
````

给你二叉搜索树的根节点 root ，该树中的恰好两个节点的值被错误地交换。请在不改变其结构的情况下，恢复这棵树 。设计算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

e.g.:
---
1.

<div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:

### 算法设计思路
因为二叉搜索树的特点是左子树所有节点值 < 根节点值 < 右子树所有节点值，所以任意一个二叉搜索树的中序遍历都是升序

本题规定树中恰好两个节点的值被错误交换，那么中序遍历的结果一定有两个元素的顺序不符合升序，也就是前一个节点值 > 当前节点值

所以算法思路是：

1.对二叉搜索树进行中序遍历；

2.用prev记录上一个访问过的节点；

3.如果发现prev.val>当前节点.val，说明这两个节点顺序有问题；

4.第一次发现错误时，记录first=prev；（第一个错误的节点是大的那个）

5.每次发现错误时，都更新second=当前节点；（第二个错误的节点是小的那个，但第二个错误的节点可能与第一个错误节点相邻也可能不相邻，所以每次发现都需更新 **ps:这条思路借助AI思考**）

6.最后交换first.val和second.val，恢复二叉搜索树。

---

### 时间复杂度计算

设二叉树中一共有$n$个节点。算法需要对整棵树进行一次中序遍历，每个节点只会被访问一次，在访问每个节点时，只进行了比较、赋值等常数操作。

所以时间复杂度为：$O(n)$.

---

### 算法实现


````python
from collections import deque


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def recoverTree(root):
    first = None
    second = None
    prev = None

    def inorder(node):
        nonlocal first, second, prev

        if node is None:
            return

        inorder(node.left)

        # 如果前一个节点值大于当前节点值，说明出现了错误顺序
        if prev is not None and prev.val > node.val:
            if first is None:
                first = prev
            second = node

        prev = node

        inorder(node.right)

    inorder(root)

    # 交换两个错误节点的值，不改变树的结构
    if first is not None and second is not None:
        first.val, second.val = second.val, first.val


def build_tree(values):
    if not values:
        return None

    root = TreeNode(values[0])
    queue = deque([root])
    index = 1

    while queue and index < len(values):
        node = queue.popleft()

        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])
            queue.append(node.left)
        index += 1

        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])
            queue.append(node.right)
        index += 1

    return root


def tree_to_list(root):
    if root is None:
        return []

    result = []
    queue = deque([root])

    while queue:
        node = queue.popleft()

        if node is None:
            result.append(None)
        else:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)

    # 删除末尾多余的 None
    while result and result[-1] is None:
        result.pop()

    return result


root = build_tree([1, 3, None, None, 2])
recoverTree(root)
print("输出为：", tree_to_list(root))

root = build_tree([3, 1, 4, None, None, 2])
recoverTree(root)
print("输出为：", tree_to_list(root))
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:
---
1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:

### 算法设计思路
因为本题是从根节点直接访问到叶子节点，所以用**深度优先遍历**来访问整棵树。

假设当前走到某个节点node，之前路径形成的数字是current。

那么当前节点对应的新数字就是：current=current*10+node.val

然后判断当前节点是不是叶子节点：如果node.left==None且node.right==None，说明当前路径已经结束，直接返回current。

如果不是叶子节点，就继续递归遍历左子树和右子树：左子树路径和+右子树路径和

---

### 时间复杂度计算
设二叉树中一共有$n$个节点。

DFS会访问每个节点一次，每个节点只进行一次简单计算：current=current*10+node.val

所以时间复杂度是：$O(n)$

---

### 算法实现

````python
from collections import deque


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def sum_numbers(root):
    def dfs(node, current):
        if node is None:
            return 0

        current = current * 10 + node.val

        # 如果是叶子节点，返回当前路径形成的数字
        if node.left is None and node.right is None:
            return current

        # 继续计算左子树和右子树
        return dfs(node.left, current) + dfs(node.right, current)

    return dfs(root, 0)


def build_tree(values):
    if not values:
        return None

    root = TreeNode(values[0])
    queue = deque([root])
    index = 1

    while queue and index < len(values):
        node = queue.popleft()

        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])
            queue.append(node.left)
        index += 1

        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])
            queue.append(node.right)
        index += 1

    return root


root1 = build_tree([1, 2, 3])
result1 = sum_numbers(root1)

print("输入：root = [1, 2, 3]，输出：{}".format(result1))


root2 = build_tree([4, 9, 0, 5, 1])
result2 = sum_numbers(root2)

print("输入：root = [4, 9, 0, 5, 1]，输出：{}".format(result2))
````
